# SPM example

- If `numba` is installed, the first run will be slower because of just-in-time compilation.

```python
from SPM import subspace_power_method
from utils import generate_lowrank_tensor
from time import time
import numpy as np

d = 20   # Dimension of the tensor
r = 120  # Rank of the tensor
n = 4    # Number of modes of the tensor

A = np.random.randn(d, r)
T = generate_lowrank_tensor(A, n=n)

start = time()
A_ = subspace_power_method(T, w_out=False)

print(time()-start)
print(np.linalg.norm(T.reshape(-1) - generate_lowrank_tensor(A_, n=n).reshape(-1)) / np.linalg.norm(T.reshape(-1)))
```

# Partial symmetric (2-1) SPM

- If `numba` is installed, the first run will be slower because of just-in-time compilation.

```python
from part_sym_SPM import spm_21sym
from utils import khatri_rao_power
from time import time
import numpy as np

m = 200   # Symmetric dimension of the tensor
n = 200   # Other dimension of the tensor
r = 100   # Rank of the tensor

A = np.random.randn(m, r)
B = np.random.randn(n, r)

T = np.dot(khatri_rao_power(A, 2), B.T).reshape(m, m, n)
start = time()
A_, B_ = spm_21sym(T, r=r, maxiter=1000, ntries=3, gradtol=1e-10, ftol=1e-5)
print("Time taken:", time() - start) 
T_ = np.dot(khatri_rao_power(A_, 2), B_.T).reshape(m, m, n)
print("Error:", np.linalg.norm(T.reshape(-1) - T_.reshape(-1)) / np.linalg.norm(T.reshape(-1)))
```
